```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105],
    'Name': ['Asha', 'Rahul', 'Priya', 'Kiran', 'Sneha'],
    'Department': ['IT', 'HR', 'Sales', 'IT', 'Sales']
})

salaries = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 106],
    'Salary': [50000, 40000, 45000, 60000]
})

display(employees)
display(salaries)
```

```python
inner_merge = pd.merge(employees, salaries, on='Emp_ID', how='inner')
display(inner_merge)
```

```python
left_merge = pd.merge(employees, salaries, on='Emp_ID', how='left')
display(left_merge)
```

```python
right_merge = pd.merge(employees, salaries, on='Emp_ID', how='right')
display(right_merge)
```

```python
outer_merge = pd.merge(employees, salaries, on='Emp_ID', how='outer')
display(outer_merge)
```

```python
df1 = pd.DataFrame({
    'Name': ['Asha', 'Rahul'],
    'Age': [22, 23]
})

df2 = pd.DataFrame({
    'Name': ['Priya', 'Kiran'],
    'Age': [21, 24]
})

combined = pd.concat([df1, df2], ignore_index=True)

display(combined)
```

```python
sales = pd.DataFrame({
    'Region': ['South', 'South', 'North', 'North', 'East', 'East'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 30000, 60000, 25000, 45000, 35000]
})

display(sales)
```

```python
pivot = pd.pivot_table(
    sales,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

display(pivot)
```

```python
wide_data = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet'],
    'January': [50000, 30000, 20000],
    'February': [55000, 35000, 25000],
    'March': [60000, 40000, 30000]
})

long_data = pd.melt(
    wide_data,
    id_vars=['Product'],
    var_name='Month',
    value_name='Sales'
)

display(long_data)
```

```python
result = (
    sales
    .groupby('Region')['Sales']
    .sum()
    .reset_index()
    .sort_values(by='Sales', ascending=False)
)

display(result)
```

```python
print("Total Sales:", sales['Sales'].sum())

print("\nAverage Sales:", sales['Sales'].mean())

print("\nSales by Region:")
display(sales.groupby('Region')['Sales'].sum())

print("\nHighest Sales:")
display(sales.loc[[sales['Sales'].idxmax()]])
```
