```python
import numpy as np
import time
```

```python
arr_1d = np.array([10, 20, 30, 40, 50])

print("1D Array:", arr_1d)
print("Dimensions:", arr_1d.ndim)
print("Shape:", arr_1d.shape)
```

```python
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("2D Array:\n", arr_2d)
print("Dimensions:", arr_2d.ndim)
print("Shape:", arr_2d.shape)
```

```python
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("3D Array:\n", arr_3d)
print("Dimensions:", arr_3d.ndim)
print("Shape:", arr_3d.shape)
```

```python
numbers = np.arange(1, 13)

print("Original Array:", numbers)

reshaped = numbers.reshape(3, 4)

print("\nReshaped Array:\n", reshaped)

print("\nFlattened Array:", reshaped.flatten())
```

```python
sales = np.array([
    [1000, 2000, 3000],
    [1500, 2500, 3500],
    [2000, 3000, 4000]
])

print("Sales Data:\n", sales)

print("\nTotal Sales:", np.sum(sales))

print("Row-wise Total:", np.sum(sales, axis=1))

print("Column-wise Total:", np.sum(sales, axis=0))

print("Average Sales:", np.mean(sales))
```

```python
marks = np.array([45, 78, 90, 32, 85, 60, 95])

passed = marks[marks >= 50]

print("Original Marks:", marks)
print("Students Passed:", passed)
```

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

selected = numbers[[0, 2, 4]]

print("Original Array:", numbers)
print("Selected Elements:", selected)
```

```python
data = np.array([45, 55, 60, 70, 75, 80, 90, 95])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("75th Percentile:", np.percentile(data, 75))
```

```python
study_hours = np.array([1, 2, 3, 4, 5, 6])
exam_marks = np.array([40, 45, 55, 65, 75, 85])

correlation = np.corrcoef(study_hours, exam_marks)

print("Correlation Matrix:\n", correlation)

print("Correlation Coefficient:", correlation[0, 1])
```

```python
size = 1_000_000

python_list = list(range(size))
numpy_array = np.arange(size)

# Python Loop
start = time.time()

python_result = [x * 2 for x in python_list]

python_time = time.time() - start

# NumPy Operation
start = time.time()

numpy_result = numpy_array * 2

numpy_time = time.time() - start

print("Python Loop Time:", python_time)
print("NumPy Execution Time:", numpy_time)

print("Performance Comparison:")
print("NumPy is faster:", numpy_time < python_time)
```

```python
comparison = {
    "Method": ["Python Loop", "NumPy"],
    "Execution Time (Seconds)": [python_time, numpy_time]
}

import pandas as pd

comparison_df = pd.DataFrame(comparison)

display(comparison_df)
```
