```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Asha', 'Rahul', 'Priya', 'Kiran', 'Sneha', 'Arjun', 'Meena', 'Ravi', 'Anita', 'John'],
    'Department': ['IT', 'hr', 'Sales', 'IT ', 'sales', 'HR', 'IT', ' sales ', 'HR', 'IT'],
    'Age': [25, 30, np.nan, 28, 150, 32, 27, 29, 26, 31],
    'Salary': [50000, 45000, 60000, np.nan, 55000, 48000, 52000, 51000, 47000, 50000],
    'Experience': [2, 5, 3, 4, 6, np.nan, 2, 3, 4, 5]
}

df = pd.DataFrame(data)

# Add a duplicate record
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

display(df)
```

```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
```

```python
df['Experience'] = df['Experience'].ffill()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicates Before Cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates After Cleaning:", df.duplicated().sum())
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[(df['Age'] < lower_limit) | (df['Age'] > upper_limit)]

print("Outliers:")
display(outliers)
```

```python
df['Age'] = df['Age'].clip(lower=lower_limit, upper=upper_limit)

print("Outliers Capped Successfully")
```

```python
df['Department'] = df['Department'].str.strip().str.title()

df['Name'] = df['Name'].str.strip().str.title()

display(df)
```

```python
comparison = pd.DataFrame({
    'Metric': ['Total Rows', 'Missing Values', 'Duplicate Rows'],
    'Before Cleaning': [11, 3, 1],
    'After Cleaning': [
        len(df),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

display(comparison)
```

```python
print("Final Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

display(df.head(10))
```

```python
df.to_csv('cleaned_dataset.csv', index=False)

print("Cleaned dataset exported successfully!")
```
