```python
import pandas as pd
```

```python
data = {
    "Product": ["Laptop", "Phone", "Laptop", "Headphones", "Phone", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Accessories", "Electronics", "Electronics"],
    "Region": ["South", "North", "South", "West", "North", "East"],
    "Sales": [50000, 20000, 55000, 3000, 25000, 48000],
    "Quantity": [2, 3, 2, 5, 4, 1]
}

df = pd.DataFrame(data)

df
```

```python
df_clean = df.copy()

df_clean.loc[0, "Sales"] = None

df_clean.isnull().sum()
```

```python
df_clean["Sales"] = df_clean["Sales"].fillna(
    df_clean["Sales"].mean()
)

df_clean.isnull().sum()
```

```python
df.groupby("Region")["Sales"].sum()
```

```python
df.groupby("Category").agg(
    total_sales=("Sales", "sum"),
    average_sales=("Sales", "mean"),
    total_quantity=("Quantity", "sum")
)
```

```python
df_clean.to_csv("cleaned_sales.csv", index=False)
```
